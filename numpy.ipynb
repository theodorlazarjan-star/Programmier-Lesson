```python
# Problem: Se da unf feature matrix X de shape (N, D) . Fa StandardScaler pe matricea aceasta astfel incat fiecare coloana sa aiba o medie de 0 si o deviatie standard de 1

def StandardScaler(X):
  mean = np.mean(X, axis = 0)
  std = np.std(X, axis = 0)

  return (X - mean) / std


data = np.array([ [10.0, 20.0], [20.0, 400.0], [30.0, 600.0] ])
normalized_data = StandardScaler(data)
```
