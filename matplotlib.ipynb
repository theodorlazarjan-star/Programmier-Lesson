```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, y, label = "sin(x)", color = "red")
```

```python
# Add annotations si labele
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, y, label = "sin(x)", color = "navy")

ax.set_title("Basic sine wave")
ax.set_xlabel("X_axis")
ax.set_ylabel("Y_axis")
ax.grid(True, linestyle="--", alpha = 0.6)
ax.legend()

plt.show()
```

```python
categories = ['A', 'B', 'C']
values = [23, 45, 56]
data = np.random.randn(1000)

fix, axes = plt.subplots(1, 3, figsize = (15, 4))

# Bar chart
axes[0].bar(categories, values, color = "red", edgecolor = "black")
axes[0].set_title("BarChart")

# Scatter plot
data_for_scatter = data[:len(x)] # Slice data to match the length of x
axes[1].scatter(x, data_for_scatter, color = "blue", edgecolor = "black")
axes[1].set_title("ScatterPlot")

# Histogram
axes[2].hist(data, bins = 20, color = "green", edgecolor = "black")
axes[2].set_title("Histogram")

plt.tight_layout()
plt.show()
```
