```python
import pandas as pd

data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics", "Accessories"],
    "Price": [1200, 25, 75, 300, None],
    "Stock": [15, 120, 45, 30, 80]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# Inspect & filter

mask = (df["Category"] == "Electronics") & (df["Stock"] > 3)
df[mask]
```

```python
# Handle missing data
df["Price"] = df["Price"].fillna(50.0)
df
```

```python
# Calculate the total sum Stock available for each category
# Aggregation

total_stock_by_category = df.groupby("Category")["Stock"].sum()
total_stock_by_category
```

```python
# Sorting

sorted_df = df.sort_values(by="Price", ascending=False)
sorted_df
```
